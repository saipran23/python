```python
def divide(a,b):
    try:
        res = a/b
    except ZeroDivisionError as e:
        print("Error: Cannot divide by zero.")
        return None
    finally:
        print("Execution completed.")
    return res

divide(10, 0)  
```

```python
def read_file(filename):
    try:
        file = open(filename)
        content = file.read()

    except FileNotFoundError as e:
        print(f"Error: {filename} not found.")
        return None
    finally:
        if 'file' in locals() and not file.closed:
            file.close()
            print("File closed.")
    return content
read_file("example.txt")
```

```python
def sum_list(numbers):
    total = 0;
    try:
        for item in numbers:
            total += item
    except TypeError as e:
        print("Error: All items in the list must be numbers.")
        total = None
    finally:
        print("Execution completed.")
    return total
```

```python
def get_dict_key(d, key):
    try:
        value = d[key]
    except KeyError as e:
        print(f"Error: Key '{key}' not found in dictionary.")
        return None
    finally:
        print("Execution completed.")   
    return value
```

```python
def nested(s):
    try:
        try:
            num = int(s)
        except ValueError as e:
            print(f"Error: Cannot convert '{s}' to an integer.")
        finally:
            print("Inner try-except block completed.")
        
        if num is not None:
            
            try:
                result = 10 / num
            except ZeroDivisionError as e:
                print("Error: Cannot divide by zero.") 
            finally:
                print("Inner try-except block completed.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
    finally:
        print("Outer try-except block completed.")
        
nested("0")
```

```python
def risky_function():
    raise ValueError("An error occurred in risky_function.")

def handle_risky_function():
    try:
       risky_function()
    except ValueError as e:
        print(f"Error: {e}")
    finally:
        print("Execution completed.")
    
handle_risky_function()
```

```python
class NegativeNumberError(Exception):
    """ArithmeticError: Negative numbers are not allowed."""

def check_for_negatives(lst):
    try:
        for num in lst:
            if num < 0:
                raise NegativeNumberError(f"Negative number found: {num}")
    except NegativeNumberError as e:
        print(f"Error: {e}")
    finally:
        print("Execution complete.")
  
check_for_negatives([1, -2, 3, 4])      
```
