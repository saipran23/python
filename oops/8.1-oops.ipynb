Classes And Objects

```python
class car:
    
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
    
    def display(self):
        return f"{self.year} {self.make} {self.model}"
    
    
tata = car("Tata", "Nexon", 2021)
print(tata.display())
```

```python
## Modding a bank account 


class BankAccount:
    
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
    def checkBalance(self):
        return f"{self.name} has a balance of {self.balance}"
    
    def deposit(self, amount):
        self.balance += amount
        return f"{self.name} has deposited {amount}. New balance is {self.balance}"
    def withdraw(self, amount):
        if amount > self.balance:
            return f"Insufficient funds. {self.name} has only {self.balance}"
        else:
            self.balance -= amount
            return f"{self.name} has withdrawn {amount}. New balance is {self.balance}"
        

account = BankAccount("Reddy", 1000)
account.checkBalance()
account.withdraw(100)
    
    
```
