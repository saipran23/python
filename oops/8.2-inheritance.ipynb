Inheritance:

```python
## parent class

class Car:
    def __init__(self, windows, doors, color):
        self.windows = windows
        self.doors = doors
        self.color = color
        
    def drive(self):
        print("The car is driving.")
        
car1 = Car(4, 4, "red")
car1.drive()
```

```python
class Tata(Car):
    def __init__(self, windows, doors, color, model):
        super().__init__(windows, doors, color)
        self.model = model
        
    def selfDrive(self):
        print(f"The Tata {self.model} car is self-driving.")
        super().drive()
        
        

tata1 = Tata(4, 4, "blue", "Nexon")
tata1.selfDrive()
    


    
```

```python

class a:
    def __init__(self, name):
        self.name = name

class b:
    def __init__(self, name, age):
        super().__init__(name)
        self.age = age
        
class c(a, b):
    def __init__(self, name, age, gender):
        a.__init__(self, name)
        b.__init__(self, age)
        self.gender = gender
        
    def display(self):
        print(f"Name: {self.name}, Age: {self.age}, Gender: {self.gender}")
```
