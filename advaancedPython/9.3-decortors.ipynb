```python
## clousure

def main_welcome(func):
    msg = "Welcome to the program!"
    def inner_welcome():
        print("welcome to the inner method")
        func("in the print function")
        print(msg)
    return inner_welcome()

main_welcome(print)
```

```python
def main_welcome(func):
    msg = "Welcome to the program!"
    def inner_welcome():
        print("welcome to the inner method")
        func()
        print(msg)
    return inner_welcome()
```

```python
def introduction():
    print("This is the introduction function.")
    

introduction()
```

```python
main_welcome(introduction)
```

```python
@main_welcome
def introduction():
    print("This is the introduction function.")
```
