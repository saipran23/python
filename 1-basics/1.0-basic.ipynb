```python
## Basic syntax Ru;les for Python
## Case Sensitivity: Python is case sensitive. For example, the variable name myVariable and myvariable are considered different variables.
## Indentation: Python uses indentation to define blocks of code. Consistent indentation is crucial for the correct execution of the program.
```

```python
## Line Continuation: If a statement is too long to fit on a single line, you can use the backslash (\) character to indicate that the statement continues on the next line. 
## Alternatively, you can use parentheses, brackets, or braces to enclose the statement and break it across multiple lines without using a backslash.

total = 1 + 2 + 3 + \
        4 + 5 + 6
print(total)  
```

```python
## Multiple Statements on a Single Line: You can write multiple statements on a single line by separating them with semicolons (;). However, it is generally recommended to use separate lines for better readability.
x = 5; y = 10; z = 15
print(x, y, z)
```

```python
## Semantic Rules: Python has certain semantic rules that govern the meaning and behavior of the code. For example, variable names must follow specific naming conventions, and certain keywords have predefined meanings in the language.

## variable assignment
my_variable = 10 ## the kernal automaticall understand the age is integer and assign the value to the variable my_variable

## Type Inference: Python uses dynamic typing, which means that the type of a variable is inferred based on the value assigned to it. You don't need to explicitly declare the type of a variable.
type(my_variable) ## this will return the type of the variable my_variable
```
