SETS


Sets are a built-in data type in Python used to store collections of unique items. They are unordered, meaning
that the elements do not follow a specific order, and they do not allow duplicate elements. Sets are useful for
membership tests, eliminating duplicate entries, and performing mathematical set operations like union,
intersection, difference, and symmetric difference.

```python
my_set = {1, 2, 3, 4}
print(my_set, type(my_set))
```

```python
st = set()
print(st, type(st))
```

```python
# Adding and removing a elements

my_set.add(5)
print(my_set)

my_set.remove(2)
print(my_set)
```

```python
# in normal remove() method we the elemet is not present in the set it will throw an error. 
# But in discard() method it will not throw an error if the element is not present in the set.

my_set.discard(10)
print(my_set)
```

```python
# pop()- remove the random element from the set and return it. If the set is empty, it will raise a KeyError.   
removed_element = my_set.pop()
print("Removed element:", removed_element)
print("Set after pop:", my_set)
```

```python
## clear all elements from the set using clear() method. It will remove all the elements from the set and make it empty.
my_set.clear()
```

```python
## Set MemeberShip test

my_set = {1, 2, 3, 4, 5}
print(3 in my_set)  # True
print(6 in my_set)  # False
```

```python
## Math operations on sets

set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

##union 

union_set = set1.union(set2)
print("Union:", union_set)


# Intersection of two sets  

intersection_set = set1.intersection(set2)
print("Intersection:", intersection_set)

set1.intersection_update(set2)
print("Set1 after intersection_update:", set1)
```

```python

set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}

## difference of two sets

set1.difference(set2)  # {1, 2, 3}
print("Difference:", set1.difference(set2))
```

```python
## Symmetric difference of two sets - common elememts are removed from both the sets and the remaining elements are returned as a new set.

set1.symmetric_difference(set2)  # {1, 2, 3, 7, 8, 9}
```

```python
## Set Methods

set1 = {1, 2, 3, 4, 5, 6, 7, 8}
set2 = {4, 5, 6, 7, 8}

# is Subset - check if set1 is a subset of set2. It will return True if all the elements of set1 are present in set2, otherwise it will return False.

print("Is set1 a subset of set2?", set1.issubset(set2))  # False

# is Superset - check if set1 is a superset of set2. It will return True if all the elements of set2 are present in set1, otherwise it will return False.

print("Is set1 a superset of set2?", set1.issuperset(set2))  # False   
```
