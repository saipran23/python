DICTINARIES


Dictionaries are unordered collections of items. They store data in key-value pairs. Keys must be unique and
immutable (e.g., strings, numbers, or tuples), while values can be of any type.

```python
## creating a dictionary

empty_dist = {}
print(type(empty_dist))  # Output: <class 'dict'>
```

```python
empty_dist = dict()
print(type(empty_dist))  # Output: <class 'dict'>
```

```python
student = {"name": "John", "age": 25, "courses": ["Math", "CompSci"]}

print(student)  # Output: {'name': 'John', 'age': 25, 'courses': ['Math', 'CompSci']}
```

```python
#Error - single key alwasy use.

student = {"name": "John", "name": 25, "courses": ["Math", "CompSci"]}
print(student)
```

```python
#Accessing dict elements

student = {"name": "John", "age": 25, "courses": ["Math", "CompSci"]}

print(student["name"])
print(student["age"])

# accessing  using a default get() method

print(student.get("age"))
print(student.get("phone")) 
print(student.get("phone", "Not Found")) # default value if key not found 
```

```python
# Modifying dict elements - mutable. ( key is unique)

print(student)

student["age"] = 30 # updated the age 
student["phone"] = "555-5555"  # adding a new key-value pair

print("Updated student:", student)
```

```python
# deleting dict elements

del student["age"]  # delete the age key-value pair
print("After deleting age:", student)
```

```python
## Dictionary methods

keys = student.keys()
print("Keys:", keys)
values = student.values()
print("Values:", values)


items = student.items() # get all key-value pairs as a list of tuples
print("Items:", items)
```

```python
# shallow copy of dict

student_copy = student
print(student_copy)

student["name"] = "sai"  # modifying the original dictionary
print(student_copy)



```

```python


student_copy1 = student.copy()  # creating a shallow copy of the dictionary ( it  alloacted the new memory location)

print(student_copy1)
print(student)

student["name"] = "praneeth"  # modifying the original dictionary
print(student_copy1)  # the copy remains unchanged
print(student)  # the original dictionary is modified
```

```python
## Iterating over dict elements

# using loops to iterate over keys and values or items 

## iterating over keys

for key in student:
    print(key)
    




```

```python
## iterating over values

for value in student.values():
    print(value)
```

```python

## iterating over key-value pairs

for key, value in student.items():
    print(key, ":", value)
```

```python
## Nested Disctionaries

# A dictionary can contain another dictionary as a value.

students = {
    "student1":{
        "name": "Alice",
        "age": 20,
        "grade": "A"
    },
    "student2":{
        "name": "Bob",
        "age": 22,
        "grade": "B"
    }
}

print(students)
```

```python
## Access nested dictionary elements

print(students["student1"]["name"])
print(students["student1"]["age"])
```

```python
## Iterating over nested dictionaries

for student_id, student_info in students.items():
    print("Student ID:", student_id)
    for key, value in student_info.items():
        print(key, ":", value)
    print()  # Print a blank line between students
```

```python
## Dictionary comprehension

squares = {x : x**2 for x in range(5)}
print(squares) 
```

```python
## condidtional dictionary comprehension

even_squares = {x : x**2 for x in range(10) if x % 2 == 0}
print(even_squares)  
```

```python
## Merge 2 dictionaries

dict1 = {'a': 1, 'b': 2}
dict2 = {'b': 3, 'c': 4}

merged_dict = {**dict1, **dict2}  ## ** unpacks the key-value pairs from both dictionaries into a new dictionary. If there are duplicate keys, the values from the second dictionary will overwrite those from the first.

print(merged_dict)
```
