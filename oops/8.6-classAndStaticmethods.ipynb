```python
class BankAccount:

    bank_name = "SBI"

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    @classmethod
    def changeBankName(cls, new_name):
        cls.bank_name = new_name


a1 = BankAccount("Reddy", 1000)
a2 = BankAccount("Rahul", 2000)

print(a1.bank_name)
print(a2.bank_name)

BankAccount.changeBankName("HDFC")

print(a1.bank_name)
print(a2.bank_name)

a2.bank_name = "ICICI"
```
