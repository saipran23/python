Importing Modules in Python: Modules and Packages


```python
import math


math.sqrt(16)
```

```python
from math import sqrt, pi


print(sqrt(16))
print(pi)
```

```python
import numpy as np

np.array([1, 2, 3, 4, 5])
```

```python
from math import *

pi
```
