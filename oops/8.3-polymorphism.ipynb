```python
# method overring

class Parent:
    def show(self):
        print("This is parent class method")
        

class Child(Parent):
    def show(self):
        print("This is child class method")
        

al1 = Child()
al1.show()

al2 = Parent()
al2.show()
```

```python
## plymorphism with functions and methods

class Shape:
    def area(self):
        return "the area of shape is undefined"
    

class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth
        
    def area(self):
        return self.length * self.breadth
    
    
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
        
    def area(self):
        return 3.14 * self.radius * self.radius   
    
    
def print_area(shape):
    print(shape.area())
    

r = Rectangle(5, 10)
c = Circle(7)

print_area(r)
print_area(c)

Rectangle(3, 4).area()
    
    
```

```python
## plymorphism with functions and methods

class Shape:
    def area(self):
        return "the area of shape is undefined"
    

class Rectangle(Shape):
        
    def area(self, length, breadth):
        return length * breadth
    
    def area(self, length):
        return length * length
    
    
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
        
    def area(self):
        return 3.14 * self.radius * self.radius   
    
    
def print_area(shape):
    print(shape.area())
    

r = Rectangle()
c = Circle(7)

r.area(5, 10)  
r.area(5)

    
```

```python
# plymorphism with abstract base class

from abc import ABC, abstractmethod

class Vechicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass
    

## derived class must implement the abstract method

class Car(Vechicle):
    def start_engine(self):
        print("Car engine started")
        
        
        
class Motorcycle(Vechicle):
    def start_engine(self):
        print("Motorcycle engine started")        
```
