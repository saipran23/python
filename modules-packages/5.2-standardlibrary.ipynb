```python
## Data Serialization 


import json

data = {
    "name": "John",
    "age": 30,
    "city": "New York"
}

str = json.dumps(data)
print(str)
print(type(str))


parsed_data = json.loads(str)
print(parsed_data)
print(type(parsed_data))

```

```python
## csv

import csv

with open('source.csv', mode='w', newline='') as file:
    writter = csv.writer(file)
    writter.writerow(['Name', 'Age', 'City'])
    writter.writerow(['ppr', '99', 'kavali'])
    
with open('source.txt', mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
        
```

```python
# regular expressions
import re

pattern = r'\d+'
text = 'There are 12 apples and 34 bananas.'

match = re.search(pattern, text)
if match:
    print("Match found:", match.group())
    
```
