```python
## pass - null operation, when it is required syntactically but you do not want any command or code to execute.


for i in range(5):
    if i == 3:
        pass
    
    print(i)
```

```python
for i in range(0, 5):
    for j in range(0, 5):
        print(f"i = {i}, j = {j}")
```

```python
n = 10
sum = 0

for i in range(1, n+1):
    print(f"Adding {i} to the sum")
    sum += i
print(f"Sum of first {n} numbers is: {sum}")
```

```python
for i in range(1, 101):
    if i > 0:
        for j in range(2, i):
            if (i % j) == 0:
                break
        else:
            print(i)
print("The above numbers are prime numbers between 1 and 100")
```
