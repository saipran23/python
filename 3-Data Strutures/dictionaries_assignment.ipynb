```python
cubes = { i : i **3 for i in range(1, 11)}
print(cubes)
```

```python
odd_squares = { i : i **2 for i in range(1, 11) if i % 2 != 0}
print(odd_squares)

even_squares = { i : i **2 for i in range(1, 11) if i % 2 == 0}
print(even_squares)

merged = {**odd_squares, **even_squares}
print(merged)

even_squares.update(odd_squares)
print(even_squares)
```

```python
d = {i: i**2 for i in range(1, 11)}

even_dict = {k:  v for k, v in d.items() if k % 2 == 0}
print(even_dict)
```

```python
d = {i: i**2 for i in range(1, 11)}

swaped = {v: k for k, v in d.items()}
print(swaped)
```
