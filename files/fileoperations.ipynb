```python

with open("example.txt", "r") as file:
    for line in file:
        print(line.strip())  #.strip() removes the newline character at the end of each line
```

```python
with open("example.txt", "r") as file:
    content = file.read()
    print(content)
```

```python

## writing to a file

with open("example.txt", "w") as file:
    file.write("This is a new line.\n")
    file.write("This is another new line.\n")
```

```python
# write a file with out overwriting the existing content

with open("example.txt", "a") as file:
    file.write("This line is appended to the file.\n")
    file.write("This is another appended line.\n")
```

```python
## writing a list of lines to a file 
lines = ["Line 1\n", "Line 2\n", "Line 3\n"]
with open("example.txt", "a") as file:
    file.writelines(lines)
```

Binary Files


```python

data = b'\xDE\xAD\xBE\xEF'
with open('example.bin', 'wb') as file:
    file.write(data)
    
```

```python
with open('example.bin', 'rb') as file:
    data = file.read()
    print(data)
```

```python
## read the content from source file and write in the destination file

with open('example.txt', 'r') as source_file:
    data = source_file.read()
with open('destination.txt', 'w') as destination_file:
    destination_file.write(data)
```

```python
with open('example.txt', 'r') as file:
    lines = file.readlines()
    line_count = len(lines)
    word_count = sum(len(line.split()) for line in lines)
    char_count = sum(len(line) for line in lines)

print(f"Line count: {line_count}")
print(f"Word count: {word_count}")
print(f"Character count: {char_count}") 
```

```python
# the w+ mode opens the file for both reading and writing. If the file already exists, it will be truncated to zero length, effectively erasing its contents. If the file does not exist, a new file will be created.


with open('example.txt', 'w+') as file:
    file.write("hello world.\n")
    file.write("This is another new line.\n")
    
    ## move the file cusor to the beginning of the file before reading
    file.seek(0)
    
    content = file.read()
    print(content)
```
